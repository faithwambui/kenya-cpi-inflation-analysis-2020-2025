```python
df = df.rename(columns={
    'Frequency (M)': 'Frequency',
    'Item Price (_T)': 'Division',
    'Unit of measure (IX)': 'Unit',
    'Time Period (2019-02)': 'Date',
    'Observation': 'CPI',
    'Observation Status': 'Status'
})
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%Y-%m')
```

```python
df_filtered = df[
    (df['Date'] >= '2020-01-01') &
    (df['Date'] <= '2025-12-31')
].copy()
```

```python
print("Dataset shape:", df_filtered.shape)
print("First date:", df_filtered['Date'].min())
print("Last date:", df_filtered['Date'].max())
print("Number of divisions:", df_filtered['Division'].nunique())
```

```python
df_filtered = df_filtered.drop(columns=['Status'])
```

```python
highest_cpi = df_filtered.loc[df_filtered['CPI'].idxmax()]
lowest_cpi = df_filtered.loc[df_filtered['CPI'].idxmin()]

print("Highest CPI:")
print(highest_cpi)

print("\nLowest CPI:")
print(lowest_cpi)
```

```python
division_analysis = (
    df_filtered.groupby('Division')['CPI']
    .agg(['mean', 'min', 'max'])
    .round(2)
    .sort_values(by='mean', ascending=False)
)

division_analysis
```

```python
import matplotlib.pyplot as plt

division_analysis['mean'].sort_values().plot(
    kind='barh',
    figsize=(10, 7),
    title='Average Consumer Price Index by Division (2020–2025)',
    xlabel='Average CPI',
    ylabel='Division'
)

plt.tight_layout()
plt.show()
```

```python
start_cpi = (
    df_filtered[df_filtered['Date'] == '2020-01-01']
    .set_index('Division')['CPI']
)

end_cpi = (
    df_filtered[df_filtered['Date'] == '2025-12-01']
    .set_index('Division')['CPI']
)

growth_analysis = pd.DataFrame({
    'January 2020 CPI': start_cpi,
    'December 2025 CPI': end_cpi
})

growth_analysis['CPI Change (%)'] = (
    (growth_analysis['December 2025 CPI'] -
     growth_analysis['January 2020 CPI']) /
    growth_analysis['January 2020 CPI']
) * 100

growth_analysis = growth_analysis.sort_values(
    'CPI Change (%)', ascending=False
).round(2)

growth_analysis
```

```python
ax = division_analysis['mean'].sort_values().plot(
    kind='barh',
    figsize=(12, 8),
    title='Average CPI by Expenditure Division (2020–2025)',
    color='steelblue'
)

ax.set_xlabel('Average Consumer Price Index')
ax.set_ylabel('Expenditure Division')

for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=4)

plt.tight_layout()
plt.show()
```

```python
selected_divisions = [
    'ALL ITEMS',
    'FOOD AND NON-ALCOHOLIC BEVERAGES',
    'TRANSPORT'
]

trend_data = df_filtered[
    df_filtered['Division'].isin(selected_divisions)
]

plt.figure(figsize=(12, 6))

for division in selected_divisions:
    division_data = trend_data[
        trend_data['Division'] == division
    ]

    plt.plot(
        division_data['Date'],
        division_data['CPI'],
        label=division
    )

plt.title('Kenya CPI Trends: Food, Transport and All Items (2020–2025)')
plt.xlabel('Year')
plt.ylabel('Consumer Price Index')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
df = df.sort_values(['Division', 'Date']).copy()

df['YoY_Inflation'] = (
    df.groupby('Division')['CPI']
    .pct_change(periods=12) * 100
)
```

```python
yoy_data = df[
    (df['Date'] >= '2020-01-01') &
    (df['Date'] <= '2025-12-31')
].copy()

all_items_yoy = yoy_data[
    yoy_data['Division'] == 'ALL ITEMS'
]

all_items_yoy[
    ['Date', 'CPI', 'YoY_Inflation']
].tail(15)
```

```python
valid_yoy = all_items_yoy.dropna(
    subset=['YoY_Inflation']
)

highest_inflation = valid_yoy.loc[
    valid_yoy['YoY_Inflation'].idxmax()
]

lowest_inflation = valid_yoy.loc[
    valid_yoy['YoY_Inflation'].idxmin()
]

print("Highest inflation:")
print(highest_inflation[['Date', 'CPI', 'YoY_Inflation']])

print("\nLowest inflation:")
print(lowest_inflation[['Date', 'CPI', 'YoY_Inflation']])
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    valid_yoy['Date'],
    valid_yoy['YoY_Inflation'],
    marker='o',
    markersize=3
)

plt.title('Kenya Year-on-Year Inflation Rate (2020–2025)')
plt.xlabel('Year')
plt.ylabel('YoY Inflation (%)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
selected_divisions = [
    'ALL ITEMS',
    'FOOD AND NON-ALCOHOLIC BEVERAGES',
    'TRANSPORT',
    'HOUSING, WATER, ELECTRICITY, GAS AND OTHER FUELS'
]

plt.figure(figsize=(12, 6))

for division in selected_divisions:

    division_data = yoy_data[
        yoy_data['Division'] == division
    ].dropna(subset=['YoY_Inflation'])

    plt.plot(
        division_data['Date'],
        division_data['YoY_Inflation'],
        label=division
    )

plt.title('Year-on-Year Inflation by Selected CPI Divisions (2020–2025)')
plt.xlabel('Year')
plt.ylabel('Inflation Rate (%)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
divisions_to_analyse = [
    'ALL ITEMS',
    'FOOD AND NON-ALCOHOLIC BEVERAGES',
    'TRANSPORT',
    'HOUSING, WATER, ELECTRICITY, GAS AND OTHER FUELS'
]

extreme_inflation = []

for division in divisions_to_analyse:

    data = yoy_data[
        yoy_data['Division'] == division
    ].dropna(subset=['YoY_Inflation'])

    highest = data.loc[data['YoY_Inflation'].idxmax()]
    lowest = data.loc[data['YoY_Inflation'].idxmin()]

    extreme_inflation.append({
        'Division': division,
        'Highest Inflation (%)': round(highest['YoY_Inflation'], 2),
        'Highest Month': highest['Date'].strftime('%Y-%m'),
        'Lowest Inflation (%)': round(lowest['YoY_Inflation'], 2),
        'Lowest Month': lowest['Date'].strftime('%Y-%m')
    })

extreme_inflation_df = pd.DataFrame(extreme_inflation)

extreme_inflation_df
```

```python
annual_inflation = (
    yoy_data.groupby([
        'Division',
        yoy_data['Date'].dt.year
    ])['YoY_Inflation']
    .mean()
    .reset_index()
)

annual_inflation.columns = [
    'Division',
    'Year',
    'Average Inflation (%)'
]

annual_inflation
```

```python
annual_pivot = annual_inflation.pivot(
    index='Division',
    columns='Year',
    values='Average Inflation (%)'
).round(2)

annual_pivot
```

```python
all_items_annual = annual_inflation[
    annual_inflation['Division'] == 'ALL ITEMS'
].copy()

all_items_annual.sort_values(
    'Average Inflation (%)',
    ascending=False
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

plt.imshow(
    annual_pivot,
    aspect='auto',
    cmap='YlOrRd'
)

plt.colorbar(label='Average Inflation (%)')

plt.xticks(
    range(len(annual_pivot.columns)),
    annual_pivot.columns
)

plt.yticks(
    range(len(annual_pivot.index)),
    annual_pivot.index
)

plt.title('Annual Average Inflation by CPI Division (2020–2025)')
plt.xlabel('Year')
plt.ylabel('CPI Division')

plt.tight_layout()
plt.show()
```

```python
highest_by_year = annual_inflation.loc[
    annual_inflation.groupby('Year')['Average Inflation (%)'].idxmax()
]

highest_by_year.sort_values('Year')
```

```python
plt.figure(figsize=(13, 9))

plt.imshow(
    annual_pivot,
    aspect='auto',
    cmap='YlOrRd'
)

plt.colorbar(label='Average Inflation (%)')

plt.xticks(
    range(len(annual_pivot.columns)),
    annual_pivot.columns
)

plt.yticks(
    range(len(annual_pivot.index)),
    annual_pivot.index
)

for i in range(len(annual_pivot.index)):
    for j in range(len(annual_pivot.columns)):
        value = annual_pivot.iloc[i, j]

        plt.text(
            j, i, f'{value:.1f}',
            ha='center',
            va='center',
            fontsize=8
        )

plt.title('Annual Average Inflation by CPI Division (2020–2025)')
plt.xlabel('Year')
plt.ylabel('CPI Division')

plt.tight_layout()
plt.show()
```
